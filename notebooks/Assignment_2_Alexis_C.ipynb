**Question 3: Inventory Management**

You manage a small store with the following inventory:

```
inventory = [
    {'item': 'Apple', 'quantity': 50, 'price': 0.5},
    {'item': 'Banana', 'quantity': 20, 'price': 0.3},
    {'item': 'Orange', 'quantity': 30, 'price': 0.7},
    {'item': 'Grapes', 'quantity': 1, 'price': 1.5},
    {'item': 'Mango', 'quantity': 10, 'price': 2.0},
]

# A) Identify items that need restocking (quantity less than 20)
restock_items = [item['item'] for item in inventory if item['quantity'] < 20]
print(f"Items that need restocking: {restock_items}")

# B) Calculate the total value of the current inventory
total_value = sum(item['quantity'] * item['price'] for item in inventory)
print(f"Total value of current inventory: ${total_value:.1f}")

# C) Increase the price of items with quantity less than 15 by 10%
print("Updating prices...")
for item in inventory:
    if item['quantity']  < 15:
        old_price = item['price']
        item['price'] = round(old_price * 1.1, 2)
        print(f"Updated {item['item']}'s price from ${old_price} to ${item['price']}")
```

**Question 6: Matrix Addition**

Consider the following two 3x3 matrices represented as lists of lists:


```
matrix_a = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix_b = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]

# A) Add the two matrices and store the result in a new matrix
result_matrix = [
    [matrix_a[i][j] + matrix_b[i][j] for j in range(len(matrix_a[0]))]
    for i in range(len(matrix_a))
]

# B) Print the resulting matrix
print("Resulting Matrix after addition:")
for row in result_matrix:
    print(row)
```

**Question 7: Fetching Random Jokes**

```
import requests
import json

# A) Fetch a random joke using the requests package
url = "https://official-joke-api.appspot.com/random_joke"
try:
    response = requests.get(url)
    response.raise_for_status()
    joke_data = response.json()

    # The API returns 'setup' for two-part jokes, or occasionally 'joke' for single jokes
    joke_setup = joke_data.get('setup') or joke_data.get('joke')
    punchline = joke_data.get('punchline')

    # B) Print the joke and punchline
    print(f"Joke: {joke_setup}")
    print(f"Punchline: {punchline}")

    # C) Save the joke to a text file called jokes.txt
    with open("jokes.txt", "w", encoding="utf-8") as file:
        file.write(f"Joke: {joke_setup}\n")
        file.write(f"Punchline: {punchline}\n")
    print("\nJoke successfully saved to jokes.txt.")

except Exception as e:
    print(f"An error occurred: {e}")
```
