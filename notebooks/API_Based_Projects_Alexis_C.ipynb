# **Project 2**

Currency exchange

```
import sqlite3
import time

def create_database(db_name):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    # Create 'facts' table for storing the random facts
    # Each table design will be unique based on the API you visit
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS rates (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            fromCurrency TEXT NOT NULL,
            toCurrency TEXT NOT NULL,
            rate REAL NOT NULL,
            timestamp TEXT
        )
    ''')

    conn.commit()
    return conn, cursor

def insert_rate_to_db(cursor, fromCur, toCur, toCurVal, conn):
    # SQL command to insert data into the facts table
    cursor.execute('''
        INSERT INTO rates (fromCurrency, toCurrency, rate, timestamp)
        VALUES (?, ?, ?, ?)
    ''', (
        fromCur,
        toCur,
        toCurVal,
        time.time()
    ))
    conn.commit()
    print("Fact inserted successfully")

def read_rates_from_db(cursor):
    cursor.execute('SELECT * FROM rates')
    rates = cursor.fetchall()
    return rates
```

```
import requests

fromCur = input("What is your source currency?")
toCur = input("What currency to convert to?")

url = f"https://cdn.jsdelivr.net/npm/@fawazahmed0/currency-api@latest/v1/currencies/{fromCur}.json"

response = requests.get(url)
data = response.json()

toCurVal = data[fromCur][toCur]

conn, cursor = create_database("rates.db")

insert_rate_to_db(cursor, fromCur, toCur, toCurVal, conn)

rates_out = read_rates_from_db(cursor)

for rate in rates_out:
  print(rate)
```
