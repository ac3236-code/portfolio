```
#Calculator: Create a calculator that takes two numbers and an operator (addition, subtraction, multiplication, division) as input
# and outputs the result. (Bonus: Extend this to handle multiple calculations using a loop).

user_input = input("enter the numbers to be calculated and the operation as follow: n1, n2, operation")

n1, n2, op = user_input.split(",")

n1 = int(n1)

n2 = int(n2)

def calculator(n1, n2, op):
  if op == "+":
    return n1 + n2
  elif op == "-":
    return n1 - n2
  elif op == "*":
    return n1 * n2
  elif op == "/":
    if n2 == 0:
      return " you cant do that"
    else:
      return n1 / n2
  else:
    return "invalid operation"

calculator(n1,n2,op)
```

Problem 2

```
# Shopping Cart: Simulate a shopping cart where users can add items, remove items, and view the total cost.
# Store items and their prices in a dictionary.

# Catalog of items with pre-set prices
CATALOG = {
    "apple": 0.99,
    "pear": 0.50,
    "milk": 3.49,
    "bread": 2.50,
    "eggs": 2.99,
    "coffee": 8.99,
    "icecream": 5.99,
}


def display_catalog(catalog):
    """Displays available items and their prices."""
    print("n--- Available Items ---")
    for item, price in catalog.items():
        print(f" - {item.capitalize()}: ${price:.2f}")


def view_cart(cart, catalog):
    """Displays current items in the cart and total cost."""
    print("\n--- Your Shopping Cart ---")
    if not cart:
        print("Your cart is empty.")
        return

    total = 0.0
    for item, quantity in cart.items():
        item_price = catalog[item]
        subtotal = item_price * quantity
        total += subtotal
        print(
            f" - {item.capitalize()} x{quantity} @ ${item_price:.2f} each = ${subtotal:.2f}"
        )

    print(f"\nTotal Cost: ${total:.2f}")


def shopping_cart_simulation():
    # User's cart stores item name as key and quantity as value
    cart = {}

    while True:
        print("\n==============================")
        print("1. View Catalog")
        print("2. Add Item to Cart")
        print("3. Remove Item from Cart")
        print("4. View Cart & Total")
        print("5. Checkout & Exit")
        print("==============================")

        choice = input("Select an option (1-5): ").strip()

        if choice == "1":
            display_catalog(CATALOG)

        elif choice == "2":
            display_catalog(CATALOG)
            item = (
                input("\nEnter the name of the item to add: ").strip().lower()
            )

            if item in CATALOG:
                try:
                    qty = int(input(f"How many {item}s would you like? "))
                    if qty > 0:
                        cart[item] = cart.get(item, 0) + qty
                        print(f"Added {qty} {item}(s) to your cart.")
                    else:
                        print("Quantity must be greater than 0.")
                except ValueError:
                    print("Invalid quantity. Please enter a number.")
            else:
                print("Item not found in the catalog.")

        elif choice == "3":
            if not cart:
                print("\nYour cart is already empty.")
                continue

            view_cart(cart, CATALOG)
            item = (
                input("\nEnter the name of the item to remove: ")
                .strip()
                .lower()
            )

            if item in cart:
                try:
                    qty = int(
                        input(
                            f"How many {item}s to remove? (Current: {cart[item]}): "
                        )
                    )
                    if qty >= cart[item]:
                        del cart[item]
                        print(f"Removed all {item}s from your cart.")
                    elif qty > 0:
                        cart[item] -= qty
                        print(f"Removed {qty} {item}(s).")
                    else:
                        print("Quantity must be greater than 0.")
                except ValueError:
                    print("Invalid quantity. Please enter a number.")
            else:
                print("That item is not currently in your cart.")

        elif choice == "4":
            view_cart(cart, CATALOG)

        elif choice == "5":
            view_cart(cart, CATALOG)
            print("\nThank you for shopping with us!")
            break

        else:
            print("Invalid selection. Please choose a number between 1 and 5.")


# Run the simulation
if __name__ == "__main__":
    shopping_cart_simulation()
```

```
# Create a two-player Rock-Paper-Scissors game that runs in the terminal.
# The game should handle player input, check for a win condition, and declare a winner or tie.

p1_input = input("player 1: choose rock, paper, or scissors.").lower()
p2_input = input("player 2: choose rock, paper, or scissors.").lower()

valid_inputs = ["rock", "paper", "scissors"]

def rps(p1, p2):
  if p1_input not in valid_inputs or p2_input not in valid_inputs:
    return "invalid input"

  if p1_input == p2_input:
    return "its a tie"
    # p1 win cons
  if p1_input == "rock" and p2_input == "scissors":
    return "P1 wins"
  elif p1_input == "scissors" and p2_input == "paper":
    return "P1 wins"
  elif p1_input == "paper" and p2_input == "rock":
    return "P1 wins"
    # p2 win conditions
  if p1_input == "scissors" and p1_input == "rock":
    return "P2 wins"
  elif p1_input == "paper" and p1_input == "scissors":
    return "P2 wins"
  elif p1_input == "rock" and p1_input == "paper":
    return "P2 wins"
  else:
    return "p2 wins"
rps(p1_input, p2_input)
```

```
# Create a program to generate the first n digits of the Fibonacci sequence.
# The program should take input from the user for â€˜nâ€™.

fib_length = int(input("How many digits of fib to generate?"))

def fib(n):
  seq = [0,1]

  if n == 0:
    return "none"
  elif n == 1:
    return seq[0]
  elif n == 2:
    return seq
  else:
    for i in range(2,n):
      seq.append(seq[i-2] + seq[i-1])

  return seq

print(fib(fib_length))
```
